# Classification Demonstration on the Iris Dataset

**Goal:** Train a classifier on the iris dataset and calculate various metrics.

```python
#Let's start with the imports 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
```

```python
#Load the Iris dataset
iris = sklearn.datasets.load_iris()
X = iris.data
y = iris.target

feature_names = iris.feature_names
target_names = iris.target_names


#Convert to Dataframe for readability
df = pd.DataFrame(X, columns=feature_names)
df["target"] = y

df.head()
df.describe()
df["target"].value_counts()
```

```python
#Train-test split
X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, test_size=30, stratify=y)
```

```python
#Preprocessing stage
scaler = sklearn.preprocessing.StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#First, I'll start with a basic LLR model
llr = sklearn.linear_model.LogisticRegression(max_iter=200)
llr.fit(X_train_scaled, y_train)
```

```python
#Test the models
llr_y_pred = llr.predict(X_test_scaled)
```

```python
accuracy_llr = sklearn.metrics.accuracy_score(y_test, llr_y_pred)
```
